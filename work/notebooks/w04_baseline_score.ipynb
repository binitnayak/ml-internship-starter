## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

## Rule

Pages that are old and showing declining performance should be refreshed first.

The baseline score increases when:
- Content is stale.
- Traffic trend is declining.
- Engagement is low.

The page with the highest score is ranked first in the refresh queue.

| Reason Code | Meaning |
|-------------|---------|
| STALE_CONTENT | Content is old and should be refreshed |
| DECLINING_TRAFFIC | Traffic is decreasing |
| LOW_ENGAGEMENT | User engagement is low |

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Baseline score
df["baseline_score"] = 0

# Old content
df.loc[df["content_age_days"] > 365, "baseline_score"] += 50

# Declining trend
df.loc[df["trend_direction"] == "down", "baseline_score"] += 40

# Low CTR
df.loc[df["ctr"] < 0.02, "baseline_score"] += 10

# Reason code
df["reason_code"] = "REVIEW"

df.loc[df["content_age_days"] > 365, "reason_code"] = "STALE_CONTENT"
df.loc[df["trend_direction"] == "down", "reason_code"] = "DECLINING_TRAFFIC"
df.loc[df["ctr"] < 0.02, "reason_code"] = "LOW_ENGAGEMENT"

# Action
df["action"] = "Monitor"
df.loc[df["baseline_score"] >= 50, "action"] = "Refresh"

# Rank
queue = df.sort_values("baseline_score", ascending=False)

# Save
queue.to_csv("outputs/baseline_action_score.csv", index=False)

# Show top 20
queue[["content_id","baseline_score","reason_code","action"]].head(20)
```
